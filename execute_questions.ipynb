## Init

```python
# imports
from agent.agent import BasicAgent
from try_challenge.try_one_question import try_one_question
```

## Questions

```python
answers_submit_dict = {
  "username": "string",
  "agent_code": "stringstri",
  "answers": [
  ]
}
```

```python
async def run_question(index):
    agent = BasicAgent()
    dict_run = await try_one_question(agent, index)
    task_id = dict_run['task_id']
    answer_submit = dict_run['answer_str']
    answers_submit_dict['answers'].append({
        "task_id": task_id,
        "submitted_answer": answer_submit
    })
```

### Questions

```python
tried_questions = [1, 3, 5, 6, 8, 9, 10, 11, 12, 13, 14, 17, 18, 19, 20]
```

```python
for index in tried_questions:
    await run_question(index)
    # Print the submitted answer for the current question (max 300 characters)
    print(f"Completed question {index}: {answers_submit_dict['answers'][-1]['submitted_answer'][:300]}")
```
